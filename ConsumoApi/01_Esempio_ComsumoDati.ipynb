```python
import matplotlib.pyplot as plt
import pandas as pd


UrlApi = {
    "Mese": ['Gen', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    "Vendite": [1200, 1500, 1300, 1800, 2200, 1850]
}

df = pd.DataFrame(UrlApi)
df
```

```python
# Cacolare una statistica (es. Media delle vendite).mean()

media_vendite = df['Vendite'].mean()
print(f'la media delle vendite è:  {media_vendite:.1f}')
```

## Creazione del grafico

```python
plt.plot(df['Mese'], df["Vendite"], marker='.', color='red')
plt.title("Andamento delle Vendite di Filippo")
plt.xlabel("Mese")
plt.ylabel("Euro (€)")
plt.grid(True)


plt.show()
```

```python


plt.hist(df["Vendite"], bins=5, color='blue', edgecolor='#fff')
plt.title("Histogramie delle Vendite")
plt.xlabel("Facia di valore (€)")
plt.ylabel("Frequnza (Qunate volte si verifica)")
plt.grid(True)


plt.show()
```
