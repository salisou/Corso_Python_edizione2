```python
import pandas as pd

# Creazione del dataframe(dati) = dataset(Tabella)
data = {
    "nome": ["Filippo", "Salomon", "Asia", "Ana"],
    "cognome": ["Battaglia", "Rossi", "Verdi", "Neri"],
    "voto": [8, 10, 9, 7]
}

df = pd.DataFrame(data) # chiama pandas e crea un dataframe con i dati 

df
```

Fitrare dati

```python
df[df["voto"] > 8]
```

Calcolare media

```python
df["voto"].mean()
```

Aggiungere una colonna

```python
df["Promosso"] = df["voto"] > 8
df
```

Ordinare

```python
df.sort_values(by="voto", ascending=False)
df

```
