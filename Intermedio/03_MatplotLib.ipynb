# 2 Grafico a barre Ragruppate (MatploLib)

```python
import matplotlib.pylab as plt
import numpy as np

# Creazione dei dati
corsi = ['Matematica', 'Informatica', 'Fisica', 'Englese']
voti_classe_A = [7.5, 8.2, 6.8, 10]
voti_classe_B = [6.5, 9.0, 7.5, 8.2]
voti_classe_C = [6.5, 6.8, 10, 8.2]

# Posizioni delle etichette
x = np.arange(len(corsi))
width = 0.35 # Larghezza della barra

# Confronta le classe
fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, voti_classe_A, width, label="Classe A") 
rects2 = ax.bar(x + width/2, voti_classe_B, width, label="Classe B") 
rects3 = ax.bar(x + width/2, voti_classe_C, width, label="Classe C") 


ax.set_title('Confronto voti tra classi')
ax.set_ylabel('Voto Medio')
ax.set_xticks(x)
ax.set_xticklabels(corsi)
ax.legend()

plt.show()
```

#### Heatmap (Seaborn)
- Utile: per visualizza le presenze degli studeti durante le settimane o altro...

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Dati: Righe = Studenti, Colonne = Giorni della settimana (1 = Presente, 0 = Assente)
calendario = {
    'Lun': [1, 1, 0, 1, 1, 1, 0, 1, 0, 1],
    'Mar': [1, 0, 1, 1, 0, 0, 1, 1, 1, 0],
    'Mer': [0, 1, 1, 0, 1, 0, 1, 1, 0, 1],
    'Gio': [1, 1, 1, 0, 1, 1, 0, 0, 1, 1],
    'Ven': [1, 0, 0, 1, 0, 0, 1, 1, 0, 0]
}
studenti = ['Giorgio', 'Mattia', 'Ana', 'Catalina', 'Eli', 'Emma', 'Massimo', 'Alessio']
df = pd.DataFrame(calendario, index=studenti)

sns.heatmap(df, annot=True, cmap="YlGnBu")
plt.title('Registro Presenze Settimanale')
plt.show()
```

### 5. Grafico Radar (Matplotlib + Numpy)
Utile per: Valutare le competenze di un singolo studente in diverse aree.

```python
import matplotlib.pyplot as plt
import numpy as np 

# 1. Definiamo le competenze (assicurati che il numero di etichette sia uguale al numero di voti!)
labels = np.array(['Coding', 'Teoria', 'Progetto', 'Soft Skills', 'Data Analyst', 'Presenza', 'Reverse Eng'])
stats = np.array([6, 8, 10, 11, 6, 7, 4]) # 7 valori per 7 etichette

# 2. Calcolo degli angoli
angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()

# 3. Chiusura del cerchio (fondamentale per non lasciare il grafico aperto)
stats = np.concatenate((stats, [stats[0]]))
angles = np.concatenate((angles, [angles[0]]))

# 4. Creazione della figura con proiezione POLARE
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Disegna l'area colorata (fill) e il contorno (plot)
ax.fill(angles, stats, color='red', alpha=0.25)
ax.plot(angles, stats, color='red', linewidth=2)

# 5. CORREZIONE ETICHETTE
ax.set_yticklabels([])  # Nascondiamo i numeri dei cerchi interni (1, 2, 3...) per pulizia
ax.set_xticks(angles[:-1]) # Posizioniamo i nomi sulle coordinate corrette degli angoli
ax.set_xticklabels(labels) # Scriviamo effettivamente i nomi delle competenze

plt.title('Profilo competenze: Alessio', size=15, color='red', y=1.1)

plt.show()
```
