# 2 Grafico a barre Ragruppate (MatploLib)
Utile per: Confrontare i voti modi di due diversi classi (Classe A e Classe B)

```python
import matplotlib.pylab as plt
import numpy as np

# Creazione dei dati
corsi = ['Matematica', 'Informatica', 'Fisica', 'Englese']
voti_classe_A = [7.5, 8.2, 6.8, 10]
voti_classe_B = [6.5, 9.0, 7.5, 8.2]

# Posizioni delle etichette
x = np.arange(len(corsi))
width = 0.35 # Larghezza della barra

fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, voti_classe_A, width, label="Classe A") 
rects2 = ax.bar(x + width/2, voti_classe_B, width, label="Classe B") 

plt.show()
```

```python
import matplotlib.pylab as plt
import numpy as np

# Creazione dei dati
corsi = ['Matematica', 'Informatica', 'Fisica', 'Englese']
voti_classe_A = [7.5, 8.2, 6.8, 10]
voti_classe_B = [6.5, 9.0, 7.5, 8.2]

# Posizioni delle etichette
x = np.arange(len(corsi))
width = 0.35 # Larghezza della barra

# Confronta le classe
fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, voti_classe_A, width, label="Classe A") 
rects2 = ax.bar(x + width/2, voti_classe_B, width, label="Classe B") 


ax.set_title('Confronto voti tra classi')
ax.set_ylabel('Voto Medio')
ax.set_xticks(x)
ax.set_xticklabels(corsi)
ax.legend()

plt.show()
```

#### Heatmap (Seaborn)
- Utile: per visualizza le presenze degli studeti durante le settimane o altro...

```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns


# Dati: Righe studenti, colonne = giorni della settimane (1 = Presente, 0 = Assete)
data = {
    'Lun': [1, 1, 0, 1],
    'Mar': [0, 1, 1, 0],
    'Mer': [1, 1, 0, 1],
    'Gio': [0, 0, 1, 0],
    'Ven': [1, 1, 0, 0]
}

studenti = ['Giorgio', 'Mattia', 'Ana', 'Catalina', 'Eli']
df = pd.DataFrame(data, index=studenti)
sns.heatmap(df, annot=True, cmap='Ylgnbu')

plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Dati: Righe = Studenti, Colonne = Giorni della settimana (1 = Presente, 0 = Assente)
data = {
    'Lun': [1, 1, 0, 1],
    'Mar': [1, 0, 1, 1],
    'Mer': [0, 1, 1, 1],
    'Gio': [1, 1, 1, 0],
    'Ven': [1, 0, 0, 1]
}
studenti = ['Mario', 'Luigi', 'Anna', 'Sofia']
df = pd.DataFrame(data, index=studenti)

sns.heatmap(df, annot=True, cmap="YlGnBu")
plt.title('Registro Presenze Settimanale')
plt.show()
```
